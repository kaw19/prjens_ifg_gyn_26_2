# 1. Introdução<a id="introducao"> </a>

Vamos usar dados como **fonte de experiência** para melhorar o desempenho de algoritmos que podem aprender com dados anteriores, ou seja, vamos usar a **Aprendizagem de Máquina** (*Machine Learning*), um subcampo da **Inteligência Artificial**.

Qualquer problema de aprendizado de máquina pode ser representado pelos três conceitos a seguir:  
* Aprender a resolver uma tarefa **T**. Por exemplo, construir um filtro de *spam* que aprenda a classificar *e-mails* como *spam* ou *ham*.
* Usar de alguma experiência **E** para aprender a realizar a tarefa. Normalmente, a experiência é representada por meio de um conjunto de dados. Para o filtro de *spam*, a experiência vem como um conjunto de *e-mails*, classificados manualmente por um ser humano como *spam* ou *ham*.
* Usar uma medida de desempenho **P** para saber quão bem estamos resolvendo a tarefa e também para saber se depois de fazer algumas modificações nossos resultados estão melhorando ou piorando. A porcentagem de *e-mails* que nossa filtragem de *spam* está classificando corretamente como *spam* ou *ham* podem ser **P** para nossa tarefa de filtragem de *spam*.

O suporte programático que usaremos na execução das tarefas de **Aprendizagem de Máquina** será a linguagem Python e seus pacotes (bibliotecas): **Numpy**, **Scipy**, **Matplotlib** e **Scikit-learn**.

O pacote **Scikit-learn** é uma biblioteca de código aberto que possui diversos algoritmos populares de **Aprendizado de Máquina**. O projeto do **Scikit-learn** foi iniciado em 2007 por David Cournapeau. Ele foi construído tendo como base os pacotes **Numpy** e **Scipy** (bibliotecas padrões da computação científica). Usaremos esses pacotes para incorporar dados disponíveis como fonte de experiência que serviriam para resolver diversas tarefas comuns de programação de uma forma eficaz e eficiente.

## 1.1 Operações (tarefas)
### Classificação
Identificar uma categoria a qual pertence uma nova observação.
* Aplicações: Detecção de *spam*, reconhecimento de imagem etc.
* Algoritmos: SVM, vizinhos mais próximos (k-NN), floresta aleatória, ...
### Regressão
Prever um valor contínuo para um novo exemplo/amostra.
* Aplicações: Resposta a medicamentos, preços de ações etc.
* Algoritmos: SVR, regressão de crista (*ridge*), Lasso, ...
### Agrupamento (*Clustering*)
Agrupamento automático de objetos semelhantes em conjuntos.
* Aplicações: segmentação de clientes, agrupamento de resultados de experimentos etc.
* Algoritmos: k-médias, agrupamento espectral, mudança de média, ...
### Redução de Dimensionalidade
Transformação de dados com muitas variáveis (alta dimensão) em um conjunto menor de variáveis, mantendo as informações mais importantes. Deixa o treinamento de modelos de ML mais rápido e eficiente.
* Aplicações: melhoria de modelos preditivos (reduz o tempo de treinamento de algoritmos e evita o *overfitting*), sistemas de recomendação: simplifica matrizes gigantescas de usuários e produtos (Netflix, Amazon) para sugerir itens de forma rápida. Reconhecimento facial: reduz os pixels de uma imagem para características essenciais (olhos, nariz, boca) usando técnicas como *Eigenfaces* (PCA).
* Algoritmos: PCA (Análise de Componentes Principais), LDA (Análise Discriminante Linear) etc.
### Seleção de Modelo
Comparar, validar e escolher modelos e parâmetros.
* Objetivo: Maior precisão e desempenho por meio do ajuste de parâmetros.
* Módulos: pesquisa em grade, validação cruzada, métricas.
### Pré-processamento
Extração e normalização das características. Métodos para dimensionamento, centralização, normalização, binarização etc.
* Aplicação: Transformação de dados de entrada, como texto, para uso com algoritmos de aprendizado de máquina.
* Módulos: pré-processamento, extração de recursos.

````python
from sklearn import datasets

iris = datasets.load_iris()  # carregamento do conjunto de dados de 150 instâncias (flores de Íris)
x_iris = iris.data           # larg. e compr. da sépala e pétala de cada instância 
y_iris = iris.target         # categoria (espécie da flor Íris): 0-setosa, 1-versicolor, 2-virgínica 
# Também podemos usar a sintaxe condensada:  x_iris, y_iris = iris.data, iris.target
especie = iris.target_names

print(f"Formas (linhas, colunas): características = {x_iris.shape}, categoria = {y_iris.shape}")
print("Medidas em cm (larg x compr):")
print(f"Primeira instância (cm) - sépalas: {x_iris[0][0]} x {x_iris[0][1]}, " \
      f"pétalas: {x_iris[0][2]} x {x_iris[0][3]}, espécie: {y_iris[0]} ({especie[y_iris[0]]})")
````

<span style="float: left; font-size: 2em; line-height: 1; margin-right: 0.5em; color: red">Obs.:</span>  

O atributo `.shape` do **NumPy** retorna uma **tupla** com uma vírgula para *arrays* unidimensionais — por exemplo, `(5,)`. Em Python, os parênteses `()` sozinhos servem apenas para agrupar expressões, como em operações matemáticas: `(2 + 3) * 4`. Se o **NumPy** retornasse apenas `(5)`, o Python interpretaria isso como o número inteiro 5 entre parênteses, e não como uma **tupla**. Para que o Python entenda que se quer criar uma **tupla** de um único elemento, a inclusão da vírgula após o elemento único é **obrigatória**.

## 2. Modelo de Classificação Linear<a id="linear"></a>
Vamos começar com um problema simples de **Aprendizado de Máquina** (ML): queremos prever as espécies de flores da íris usando apenas **dois atributos**: largura e comprimento da sépala. Este é um exemplo de **problema de classificação**, no qual queremos atribuir um rótulo (um elemento retirado de um conjunto discreto definido) para um item de acordo com suas características.

### 2.1 Conjunto de Treinamento
Trata-se de um **subconjunto da amostra original**, representado pelos dois atributos que selecionamos e seus respectivos valores alvo. 

Depois de importar o conjunto de dados, selecionaremos aleatoriamente cerca de 75% das instâncias e reservaremos as restantes para fins de avaliação (o conjunto de dados de avaliação).

A função `train_test_split()` constrói automaticamente os conjuntos de dados de treinamento e teste, selecionando as amostras de forma aleatória. Por que não selecionar apenas os primeiros 75% das instâncias? Porque pode acontecer da ordem das instâncias dentro da amostra original ser importante e que as primeiras instâncias possam ser diferentes das últimas. Na verdade, se observarmos os conjuntos de dados **Iris**, as instâncias estão ordenadas por sua classe alvo, e isso implica que a proporção de classes 0 e 1 será maior no novo conjunto de treinamento, em comparação com a do conjunto de dados original... Sempre queremos que nossos dados de treinamento sejam uma amostra representativa da população que eles representam.

````python
from sklearn.model_selection import train_test_split
# from sklearn import preprocessing

# Pega o conjunto original com somente os dois atributos selecionados
x, y = x_iris[:, :2], y_iris        # somente informações de largura e comprimento das sépalas

# Fatia o conjunto original em subconjuntos de treinamento e de teste (25% das instâncias escolhidas aleatoriamente)
x_treino_o, x_teste_o, y_treino_o, y_teste_o = train_test_split(x, y, test_size=0.25, random_state=33)
print(f"Formas das amostras de treinamento: x = {x_treino_o.shape}, y = {y_treino_o.shape}")
````

````python
import matplotlib.pyplot as plt

cores = ['red', 'green', 'blue']
for i in range(len(cores)):
    xs = x_treino_o[:, 0][y_treino_o == i]
    ys = x_treino_o[:, 1][y_treino_o == i]
    plt.scatter(xs, ys, c=cores[i])
plt.legend(iris.target_names); plt.title("Distribuição das Instâncias de Treino no Espaço 2D")
plt.xlabel('Comprimento da Sépala (cm)'); plt.ylabel('Largura da Sépala (cm)'); plt.grid(True)
````

A função `scatter()` plota o primeiro valor da característica (largura da sépala) para cada instância versus segundo valor da característica (comprimento da sépala) e usa os valores da classe alvo (`y_treino`) para atribuir uma cor diferente para cada classe. O gráfico mostra como esses atributos contribuem para determinar a classe alvo de cada instância.

#### 2.1.1 Escalonamento das Características (*Features*)
As últimas três linhas do código anterior modificam o conjunto de treinamento em um processo geralmente chamado de **Escalonamento das Características**. Para cada característica, ele calcule a média, subtrai o valor médio do valor da característica e divide o resultado pelo desvio padrão. Após o escalonamento, cada característica terá média zero e desvio padrão unitário.   

Essa padronização de valores (que não altera a distribuição de frequência da amostra, e que pode ser verificada traçando os valores de **X** antes e depois do dimensionamento) é um **requisito** comum dos modelos de **Aprendizado de Máquina**, para evitar que características com valores grandes possam pesar muito nos resultados finais.

Vamos dar uma olhada em como as instâncias do conjunto de treinamento estão distribuídas no espaço bidimensional gerado pelo recurso de aprendizagem. O pacote `pyplot`, da biblioteca `matplotlib`, será usado para a exibição gráfica:

````python
import matplotlib.pyplot as plt

# Escalonamento padronizado das características (features scaling)
#   Padroniza as características removendo a média e dimensionando para uma variância unitária.
#   A pontuação (score) padrão de uma amostra 'x' é calculada como:  xp = (x - m)/s
#     onde: m é a média das amostras de treinamento, 
#           s é o desvio padrão das amostras de treinamento.
escalonador = preprocessing.StandardScaler().fit(x_treino_o) 
x_treino = escalonador.transform(x_treino_o)
x_teste = escalonador.transform(x_teste_o)

cores = ['red', 'green', 'blue']
for i in range(len(cores)):
    xs = x_treino[:, 0][y_treino_o == i]
    ys = x_treino[:, 1][y_treino_o == i]
    plt.scatter(xs, ys, c=cores[i])
plt.legend(iris.target_names); plt.title("Distribuição das Instâncias de Treino no Espaço 2D")
plt.xlabel('Comprimento da Sépala (escalonado)'); plt.ylabel('Largura da Sépala (escalonada)'); plt.grid(True)
````

No gráfico anterior vemos que a separação entre os pontos vermelhos (correspondentes à Iris Setosa) e os pontos verdes e azuis (correspondentes às outras duas espécies de Iris) é bastante clara, enquanto separar os pontos verdes (correspondentes à Iris Versicolor) dos azuis (correspondentes à Iris Virginica) parece uma tarefa muito difícil, dados as duas características disponíveis (larg & compr das sépalas).
Este é um cenário muito comum: uma das primeiras perguntas que queremos responder em uma tarefa de **Aprendizado de Máquina** é se o conjunto de características que estamos usando é realmente útil para a tarefa que estamos resolvendo, ou se precisamos adicionar novos atributos ou alterar nosso método.

## 3. Classificadores Lineares<a id="classificador"></a>

Olhando o gráfico anterior, parece que podemos desenhar uma reta que separe ambos conjuntos (talvez com a exceção de um ou dois pontos, que poderiam cair no lado incorreto separado pela reta). O modelo classificação linear faz exatamente essa separação binária: constrói uma reta, mais genericamente, um hiperplano no espaço das características, que separa otimamente ambas as classes alvos, e a usa como uma fronteira de decisão, ou seja, a associação à classe depende de qual lado do hiperplano a instância está.

Para implementar a classificação linear usaremos o **SGDClassifier** do pacote **scikit-learn**. O SGD (*Stochastic Gradient Descent* - Gradiente Estocástico Descendente) é um procedimento numérico muito popular usado na busca do mínimo local de uma função, que neste caso, a função de perda, que mede a que distância cada instância está do hiperplano (fronteira). O algoritmo calcula os coeficientes do hiperplano minimizando a função de perda.

### 3.1 Etapas da Modelagem

1. Criar o objeto do classificador correspondente (nesse caso, *SGDClassifier*).
1. Iniciar os parâmetros do classificador.
1. Treinar o modelo que melhor se ajuste (*fits*) aos dados de treinamento.

````python
# Importação do Classificador SGD (classificadores lineares, tais como SVM, regressão logística etc., com treinamento SGD).
from sklearn.linear_model import SGDClassifier

clf_SGD = SGDClassifier()
clf_SGD.fit(x_treino, y_treino_o)  # melhor ajuste do modelo aos dados de treinamento
````

<span style="float: left; font-size: 2em; line-height: 1; margin-right: 0.5em;">Obs.:</span> No código anterior usamos os valores padrão dos parâmetros da função `fit()`, mas esses **parâmetros** podem ser **muito importantes**, especialmente quando se enfrenta tarefas reais, onde a quantidade de instâncias (ou mesmo a quantidade de atributos) pode ser muito grande.  

A função **fit()** é provavelmente a mais importante no **scikit-learn**. Ela recebe os dados e as classes de treinamento, e constrói o classificador. Todo método de Aprendizagem de Máquina do tipo **supervisionado** no **scikit-learn** implementa esta função.  

O atributo `.coef_` do objeto `clf_SGD` (considerando apenas a primeira linha das matrizes, por enquanto), possui os coeficientes da fronteira linear e o atributo `.intercept_`, possui o ponto de intersecção da reta com o eixo vertical.

````python
print(clf_SGD.coef_, "\n\n", clf_SGD.intercept_)
````

````python
from IPython.display import Markdown

Markdown(f"Portanto, no plano real, com estes três valores, podemos traçar a reta (hiperplano), representada pela seguinte equação:  " \
         f"{clf_SGD.coef_[0][0]:+0.3f} . x1 {clf_SGD.coef_[0][1]:+0.3f} . x2 {clf_SGD.intercept_[0]:+0.3f} = 0 .")
````

Agora, dados `x1` e `x2` (valor real das características), calculamos o valor do lado esquerdo da equação: se esse valor for maior que zero, então o ponto está acima do limite de decisão, caso contrário, estará abaixo da reta. O algoritmo de previsão irá simplesmente verificar isso e prever a classe correspondente para qualquer nova flor de íris.

Mas, por que a matriz de coeficientes tem três linhas? Porque não informamos ao método a alteração da definição do problema, e ele resolveu o problema para as três classes da flor Íris (setosa, versicolor, virginica), e não um problema de decisão binária. Neste caso, o classificador converteu o problema em três problemas de classificação binária, em um cenário do tipo **um contra todos** (propos três retas que separam cada uma das classes das demais). O código a seguir desenha os três limites de decisão e nos permite saber se eles funcionam como esperado:

````python
x_min, x_max = x_treino[:, 0].min() - .5,  x_treino[:, 0].max() + .5
y_min, y_max = x_treino[:, 1].min() - .5,  x_treino[:, 1].max() + .5
xs = np.arange(x_min, x_max, 0.5)
fig, axes = plt.subplots(1, 3)
fig.set_size_inches(12, 6)
for i in [0, 1, 2]:
    axes[i].set_aspect('equal')
    axes[i].set_title('Classe '+ str(i) + '  X  resto')
    axes[i].set_xlabel('Comprimento da Sépala'); axes[i].set_ylabel('Largura da Sépala')
    axes[i].set_xlim(x_min, x_max); axes[i].set_ylim(y_min, y_max)
    # sca(axes[i])
    plt.scatter(x_treino[:, 0], x_treino[:, 1], c=y_treino_o, cmap=plt.cm.prism)
    ys = (-clf_SGD.intercept_[i] - xs * clf_SGD.coef_[i, 0]) / clf_SGD.coef_[i, 1]
    plt.subplot(1,3,i+1); plt.plot(xs, ys)
````

O primeiro gráfico mostra o modelo construído para o problema binário. Parece que a linha separa muito bem a **Iris Setosa** das demais. Para as outras duas tarefas, como esperávamos, existem vários pontos que ficam do lado errado do hiperplano. 

Agora, vamos supor que temos uma nova flor com largura de sépala de 4.7 cm e um comprimento de sépala de 3.1 cm, e queremos prever sua classe. Então aplicamos o classificador (depois da normalização). O método de previsão pega um arranjo de instâncias (neste caso, com apenas um elemento) e retorna uma lista de classes previstas:

````python
print(clf_SGD.predict(escalonador.transform([[4.7, 3.1]])))
````

Se o classificador estiver certo, esta flor de **Íris** é da categoria **Setosa**. Provavelmente, estamos prevendo uma classe a partir das três classes possíveis, mas que os modelos lineares são essencialmente binários: está faltando algo... 

Nosso procedimento de predição combina o resultado dos três classificadores binários e seleciona a classe na qual é mais confiável. Neste caso, selecionaremos a reta limite cuja distância até a instância for maior. Podemos verificar isso usando o método do classificador chamado de `decision_function()`:

````python
print(clf_SGD.decision_function(escalonador.transform([[4.7, 3.1]])))
````

## 4. Avaliação dos Resultados<a id="avaliacao"></a>

Em uma tarefa de classificação, a **precisão** de uma classe é o número de itens verdadeiros positivos (a quantidade de itens corretamente rotulados como pertencentes à classe positivo) dividido pelo número total de elementos rotulados como pertencentes à classe positivo (a soma de itens verdadeiros positivos e falsos positivos, que são itens rotulados incorretamente como pertencentes à classe positivo). A **recuperação** (*recall*) é definida como o número de itens verdadeiros positivos dividido pelo número total de elementos que realmente pertencem à classe positivo (ou seja, a soma de verdadeiros positivos e falsos negativos, que são itens que não foram rotulados como pertencentes à classe positiva, mas deveriam ter sido).

Numa tarefa de classificação, uma pontuação (*score*) de precisão igual a 1,0 para uma classe **C** significa que cada item rotulado como pertencente à classe **C** pertence de fato à classe **C** (mas não diz nada sobre o número de itens da classe **C** que não foram rotulados corretamente), enquanto uma **recordação** igual a 1,0 significa que cada item da classe **C** foi rotulado como pertencente à classe **C** (mas não diz nada sobre quantos itens de outras classes foram rotulados incorretamente como pertencentes à classe **C**).

#### Exemplo: Detector de Fumaça
Um detector de fumaça é geralmente projetado para cometer muitos erros do **Tipo I** (alertar em muitas situações quando não há perigo), porque o custo de um erro do **Tipo II** (não soar um alarme durante um incêndio grave) é proibitivamente alto. Como tal, os detectores de fumaça são projetados tendo em mente o **recall** (para capturar todos os perigos reais), mesmo dando pouco importância às perdas de **precisão** (e gerando muitos alarmes falsos). Na outra direção, a **proporção de Blackstone**, "É melhor que dez pessoas culpadas escapem do que que uma inocente sofra uma injustiça", enfatiza os custos de um erro **Tipo I** (condenar uma pessoa inocente). Como tal, o sistema de justiça criminal é orientado para a **precisão** (para a não condenação de inocentes), mesmo ao custo de perdas na *recall* (deixando mais culpados irem em liberdade).

O desempenho de um classificador é medido pela sua eficácia. A **medida de desempenho** mais simples é a **acurácia** (indica o grau de exatidão ou a proximidade entre um resultado obtido e o seu valor real, ou de referência): dado um classificador e um conjunto de dados de teste, mede-se a proporção de instâncias classificadas corretamente pelo classificador. 

A **acurácia** é uma média aritmética ponderada da **Precisão** e da **Precisão Inversa** (ponderada por **Viés**), bem como uma média aritmética ponderada de **Recordação** e **Recordação Inversa** (ponderada por **Prevalência**).

**Precisão Inversa** e **Recordação Inversa** são simplesmente a **precisão** e **recall** do problema inverso onde rótulos positivos e negativos são trocados (tanto para classes reais quanto para rótulos de predição). Taxa de verdadeiro positivo e taxa de falso positivo, ou equivalentemente **Recall** e **1-Inverse Recall**,
são frequentemente plotados uns contra os outros, como em curvas **ROC** (*Receiver Operating Characteristic* - curva característica de operação do receptor é um gráfico que mostra o desempenho de um modelo de classificador binário com valores limites variados) e fornecem um mecanismo de princípios para explorar compensações de pontos operacionais. Fora da área de informações, a aplicação de **recall**, **precisão** e **F1-score** é considerada falha, pois ignora a célula verdadeira negativa da tabela de contingência, e elas são facilmente manipuladas distorcendo as previsões.

Vamos testar a acurácia do classificador linear usando o conjunto de treinamento:

````python
from sklearn import metrics

y_treino_pred = clf_SGD.predict(x_treino)
print(f"Acurácia do modelo: {metrics.accuracy_score(y_treino_o, y_treino_pred)*100:.2f}%",end="")
print(" das instâncias do conjunto de treinamento são classificadas corretamente pelo classificador.")
````

Medir a acuidade no conjunto de treinamento é realmente uma má ideia. Construímos o modelo usando esses dados e é bem possível que o modelo se ajuste bem a eles, mas tenha um desempenho ruim com dados não vistos anteriormente, que é o propósito do modelo.

Esse fenômeno é chamado de ***overfitting***. Se medirmos com base nos dados de treinamento, nunca detectaremos um *overfitting*. Portanto, nunca meça a acurácia com base nos dados de treinamento. É por isso que se reserva parte do conjunto de dados original (a partição de teste) – queremos avaliar o desempenho em dados nunca antes vistos. Vamos verificar a acuidade novamente, agora com o conjunto de avaliação:

````python
y_pred = clf_SGD.predict(x_teste)
print(f"Acurácia do modelo: {metrics.accuracy_score(y_teste_o, y_pred)*100:.2f}%")
````

Obtivemos uma acurácia de 68% quando usamos o conjunto de testes. Normalmente, a acurácia (exatidão) do conjunto de teste é menor do que a acurácia do conjunto de treinamento, uma vez que o modelo está, na verdade, modelando o conjunto de treinamento, não o conjunto de teste. 

O objetivo sempre será produzir modelos que evitem ***overfitting*** quando treinados em um conjunto de treinamento, para que tenham poder de generalização suficiente para modelar corretamente os dados não vistos. 

A acurácia no conjunto de testes é uma boa medida de desempenho quando a quantidade de instâncias de cada classe é semelhante, ou seja, quando temos uma **distribuição uniforme** de classes. Mas se tivermos uma distribuição distorcida (digamos, 99% das instâncias pertencem a uma classe), um classificador que sempre preveja a classe majoritária poderia ter um excelente desempenho em termos de acurácia, apesar de ser um método extremamente ingênuo.

### 4.1 Funções de Avaliação
No **scikit-learn** existem diversas **funções de avaliação** mas mostraremos apenas as três mais populares: *precision*, *recall* e *F1-score* (ou medida-F). Elas pressupõem um problema de **classificação binária** e **duas classes** – uma positiva e outra negativa. No nosso exemplo a classe positiva poderia ser **Iris Setosa**
enquanto as outras duas classes serão combinadas em uma classe negativa. 
* **Precisão**: calcula a proporção de instâncias previstas como positivas que foram avaliadas corretamente, medindo o quão correto o classificador está quando diz que uma instância é positiva (número de resultados verdadeiros positivos dividido pelo número de todas as amostras previstas como positivas, incluindo aqueles não identificados corretamente). A **precisão** também é conhecida como **valor preditivo positivo**.
* **Recall** (recordação): conta a proporção de instâncias positivas que foram avaliadas corretamente, medindo o quão correto o classificador está quando confrontado com uma instância positiva (número de resultados positivos verdadeiros dividido pelo número de todas as amostras que deveriam ter sido identificadas como positivas). A **recordação** também é conhecida como **sensibilidade** na classificação binária diagnóstica.
* **F1-score**: esta métrica é uma média harmônica entre **precision** e **recall**, combinando ambas em um único número. Na análise estatística de sistemas de classificação binária, o **F-score** ou **F-measure** é uma medida de desempenho preditivo.  

As métricas mostradas anteriormente (precisão, recall, F1-score), podem ser definidas em termos de quatro componentes: **Verdadeiro** e **Falso**, e **Positivo** e **Negativo**.

Por exemplo, imagine que o modelo tenta prever se um e-mail é *spam* (positivo) ou não *spam* (negativo):
* Verdadeiro Positivo (**VP**/TP): O modelo previu que era *spam*, e realmente era *spam* (acerto).
* Verdadeiro Negativo (**VN**/TN): O modelo previu que não era *spam*, e realmente não era *spam* (acerto).
* Falso Positivo (**FP**): O modelo previu que era *spam*, mas não era (erro **Tipo I** / alarme falso).
* Falso Negativo (**FN**): O modelo previu que não era *spam*, mas era *spam* (erro **Tipo II** / deixou passar).

#### Tabela

|                           | Predição: **Positivo**   | Predição: **Negativo**   |
| :------------------------ | :--------------------:   | :--------------------:   |
| Classe Alvo: **Positivo** | Verdadeiro Positivo (VP) | Falso Negativo (FN)      |
| Classe Alvo: **Negativo** | Falso Positivo (FP)      | Verdadeiro Negativo (VN) |

Sendo `m` o tamanho da amostra (ou seja, `VP + VN + FP + FN`), temos as seguintes fórmulas:  
* Acurácia = (VP + VN) / m   --> mede a fração de instâncias relevantes dentre as instâncias recuperadas, ou seja, total de acertos dividido pelo total geral de amostras.
* Precisão = VP / (VP + FP)  --> mede a fração de todas as instâncias verdadeiras positivas que um modelo identifica corretamente, quantos dos casos previstos como positivos estavam certos.
* *Recall* (sensibilidade) = VP / (VP + FN)  --> mede a fração de instâncias relevantes que foram recuperadas, quantos dos positivos reais o modelo conseguiu detectar.
* F1-score = 2 * Precisão * *Recall* / (Precisão + *Recall*)  

````python
print(metrics.classification_report(y_teste_o, y_pred, target_names = iris.target_names))
````

#### Informações (tabela anterior)  

• O classificador obteve *precision* e *recall* iguais a 1,0 na classe **setosa**. Isso significa que 100% das instâncias classificadas como **setosa** são realmente instâncias **setosa** .  
• Por outro lado, na classe **versicolor** os resultados não são tão bons: temos uma precisão de 0.43, ou seja, apenas 43% das instâncias que são classificadas como **versicolor** são realmente instâncias **versicolor**. Além disso, para **versicolor**, temos um *recall* de 0.27, ou seja, apenas 27% das instâncias **versicolor** estão classificadas corretamente.

Outra métrica útil (especialmente para problemas multiclasse) é a **matriz de confusão**, que ajuda a entender onde o modelo acerta e erra, indo além da simples **acurácia** (que pode enganar se os dados estiverem desequilibrados). É muito usada em problemas de classificação binária (sim/não, verdadeiro/falso), mas também funciona com várias classes: em sua célula `(i, j)`, ela mostra a quantidade de instâncias da classe `i` que foram previstas como pertencentes à classe `j`. Um bom classificador acumulará os valores na diagonal da **matriz de confusão**, onde estão as instâncias classificadas corretamente.

````python
print(metrics.confusion_matrix(y_teste_o, y_pred))
````

O classificador nunca erra no conjunto de teste quando classifica flores de classe 0 (**setosa**). Mas, quando se depara com flores das classes 1 e 2 (**versicolor** e **virginica**), ele as confunde. A **matriz de confusão** nos fornece informações úteis para saber quais tipos de erros o classificador está cometendo.

### 4.2 Validação Cruzada

Como visto, o **conjunto de dados** é particionado em um **conjunto de treinamento** e um **conjunto de testes**. No entanto, particionar os dados resulta em menos instâncias para treinar e também, dependendo da partição específica que fazemos (geralmente feita aleatoriamente), podemos obter resultados melhores ou piores.  

A **validação cruzada** permite-nos evitar este caso específico, reduzindo a variância dos resultados e produzindo uma **pontuação mais realista** para os modelos adotados. 

As etapas usuais para validação cruzada **k-fold** (k dobras) são as seguintes:
1. Particione o conjunto de dados em **k** subconjuntos diferentes.
2. Crie **k** modelos diferentes treinando em **k-1** subconjuntos e testando no subconjunto restante.
3. Meça o desempenho em cada um dos **k** modelos e calcule a média.

#### Exemplo
Vamos treinar um **classificador linear**. Precisamos criar um **estimador composto** feito por um encanamento (*pipeline*) dos modelos de padronização e lineares. Com isso garantimos que cada iteração irá padronizar os dados e então treinar/testar os dados transformados.

* Classe **Pipeline**: Executa uma sequência de transformadores de dados com um preditor final opcional.  

    Essa classe aplica sequencialmente uma lista de **transformadores** para pré-processar os dados e, se desejar, concluir a sequência com um preditor final para modelagem preditiva. As etapas intermediárias do *pipeline* devem ser constituídas de transformadores, ou seja, eles devem implementar os métodos `.fit` e `.transform`. O estimador final só precisa implementar o método `.fit`. Os transformadores no **pipeline** podem ser armazenados em cache usando o argumento `memory`.

    O objetivo do *pipeline* é montar várias etapas que podem ser juntamente validadas de forma cruzada enquanto se definem parâmetros diferentes.  

    A classe **Pipeline** também é útil para simplificar a construção de modelos mais complexos que multiplicam em cadeia as transformações. 

Vamos escolher 5 dobras, que resultarão a cada vez, em um treinamento com 80% dos dados e teste com os 20% restantes. A **validação cruzada**, por padrão, usa a **acuidade** como medida de desempenho, mas poderíamos selecionar a medida passando qualquer função de pontuação (*score*) como argumento.

````python
from sklearn.model_selection import cross_val_score, KFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Cria um classificador composto feito por um encanamento de padronização e do modelo linear
clf = Pipeline( [('scaler', StandardScaler()), ('linear_model', SGDClassifier())] )

# Cria um iterador para a validação cruzada k-fold com 5 dobras
vc = KFold(n_splits=5, shuffle=True, random_state=33)

# Por padrão a pontuação (score) usada é o valor retornado pelo método '.score' do estimador (acurácia)
score = cross_val_score(clf, x, y, cv=vc)
print(score)
````

Obtivemos um vetor com as `k` pontuações. Podemos calcular a média e o erro padrão para obter um valor final:

````python
from scipy.stats import sem  # standard error mean

def mean_score(scores):
    return ("Escore médio: {0:.3f} (+/-{1:.3f})").format(np.mean(scores), sem(scores))

print(mean_score(score))
````

## 6. Outros Conceitos<a id="conceitos"></a>

### 6.1 Maldição da Dimensionalidade
Quando lidamos com muitos dados de treinamento para estimar adequadamente os parâmetros do modelo, podemos nos deparar também com grande número de parâmetros (e, consequentemente, a quantidade de dados de treinamento necessários para estimá-los adequadamente) cresceria rapidamente se adicionarmos mais características ou termos de ordem superior. Este fenômeno, presente em todos os modelos de aprendizado de máquina (ML) é chamada de **maldição da dimensionalidade**: quando o número de parâmetros de um modelo aumenta, os dados necessários para aprendê-los **crescem exponencialmente**.  

### 6.2 *Overfitting* e *Underfitting*
Esta noção está intimamente relacionada ao problema de *overfitting* mencionado anteriormente. Como nossos dados de treinamento não são suficientes, corremos o risco de produzir um modelo que poderia ser muito bom em prever a classe alvo no conjunto de dados de treinamento, mas falha consideravelmente quando confrontado com novos dados, ou seja, nosso modelo **não tem o poder de generalização**. É por isso que é tão importante avaliar/testar os modelos com base em dados não usados no treinamento.

A regra geral é que, para evitar *overfitting*, devemos preferir métodos simples (ou seja, com menos parâmetros), algo que poderia ser visto como uma instância do princípio filosófico da **Navalha de Occam**, que afirma que entre hipóteses concorrentes, a hipótese com menor número de suposições deve ser selecionada. 

No entanto, devemos também levar em conta as palavras de **Einstein**: 
    “Tudo deveria ser tão simples quanto possível, mas não mais simples”.

A maldição da **dimensionalidade** pode sugerir que mantenhamos os modelos simples, mas por outro lado, se o nosso modelo for demasiado simples corremos o risco de sofrer de *underfitting*. Problemas de *underfitting* surgem quando o modelo tem um poder de representação tão baixo que não consegue modelar os dados mesmo que tivéssemos todos os dados de treinamento desejados. Claramente temos *underfitting* quando nosso algoritmo não consegue atingir boas medidas de desempenho, mesmo quando medimos no conjunto de treinamento. Como resultado, teremos que alcançar um equilíbrio entre *overfitting* e *underfitting*. Este é um dos problemas mais importantes que teremos que resolver ao projetar nossos modelos de ML.

### 6.3 Viés (bias) e Variância
Considere um método extremo que, em uma configuração de **classificação binária**, sempre prevê a classe positiva para qualquer nova instância. Suas previsões são, trivialmente, sempre as mesmas, ou em termos estatísticos, possuem **variância nula**; mas não conseguirá prever exemplos negativos, sendo muito tendencioso para resultados positivos. Por outro lado, considere um método que preveja, para uma nova instância, a classe da instância mais próxima no conjunto de treinamento -- na verdade, esse método existe e é chamado de 1-vizinho mais próximo. Os pressupostos de generalização que este método utiliza são muito pequenos: tem um **viés** muito baixo; mas, se alterarmos os dados de treinamento, os resultados poderão mudar drasticamente, ou seja, sua **variância** é muito alta. Estes são exemplos extremos da compensação entre **viés** e **variância**. Pode ser demonstrado que, independentemente do método utilizado, se reduzirmos o **viés**, a **variância** aumentará e vice-versa.

Os **classificadores lineares** geralmente têm baixa **variância**: não importa qual subconjunto selecionamos para treinamento, os resultados serão semelhantes. Contudo, se a distribuição dos dados (como no caso das espécies versicolor e virginica) fizer com que as classes alvo não sejam separáveis ​​por um hiperplano, estes resultados estarão consistentemente errados, ou seja, o método é altamente tendencioso.

Por outro lado, o **kNN** tem **viés** muito baixo, mas alta **variância**: os resultados geralmente são muito bons para descrever dados de treinamento, mas tendem a variar muito quando treinados em instâncias diferentes das usadas no treinamento.
